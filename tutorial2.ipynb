```python
#3. Create a numpy array of ones and a numpy array of zeroes

ones_array = np.ones((3,3))
zeroes_array = np.zeros((3,3))

print("this is a 3x3 array of ones")
print(ones_array)

print("this is a 3x3 array of zeroes")
print(zeroes_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it

random6x3 = np.random.rand(6,3)
transposed_array = np.transpose(random6x3)

print("this is a random 6x3 array")
print(random6x3)

print("this is the transposed array")
print(transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

n, p, m = 4, 3, 2

array1 = np.random.rand(n, p)
array2 = np.random.rand(p, m)

dot_product = np.dot(array1, array2)

print("this is the first random array")
print(array1)

print("this is the second random array")
print(array2)

print("this is the dot product of the two arrays")
print(dot_product)
```
